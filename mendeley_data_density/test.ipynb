```python
from google.colab import drive
drive.mount('/content/drive')

import nibabel as nib
import numpy as np
import json, os, glob

root = "/content/drive/MyDrive/Mendeley_Data_Density_Entegrasyonu"

# 1) Bir örnek image/label çifti
img_path = sorted(glob.glob(f"{root}/imagesTr/*.nii.gz"))[0]
lbl_path = sorted(glob.glob(f"{root}/labelsTr/*.nii.gz"))[0]

img = nib.load(img_path).get_fdata()
lbl = nib.load(lbl_path).get_fdata()

print("image shape:", img.shape, "dtype:", img.dtype, "min/max:", img.min(), img.max())
print("label shape:", lbl.shape, "unique values:", np.unique(lbl))

# 2) Tüm train setindeki maskelerin birleşik unique değerleri (sınıf sayısını netleştirir)
all_vals = set()
for p in glob.glob(f"{root}/labelsTr/*.nii.gz")[:30]:  # ilk 30 dosya yeterli
    v = nib.load(p).get_fdata()
    all_vals.update(np.unique(v).tolist())
print("Tüm örneklerdeki maske değerleri:", sorted(all_vals))

# 3) JSON içeriği - density etiketleri neyi tutuyor?
with open(f"{root}/train_density.json") as f:
    d = json.load(f)
print(type(d), list(d.items())[:3] if isinstance(d, dict) else d[:3])

# 4) Dosya sayıları
print("Tr:", len(glob.glob(f"{root}/imagesTr/*")), "Val:", len(glob.glob(f"{root}/imagesVal/*")))
```
